````python
import os
import pandas as pd
from langchain.agents.agent_types import AgentType
from langchain_experimental.agents.agent_toolkits import create_csv_agent
from langchain_groq import ChatGroq
from langchain.memory import ConversationBufferMemory
from langchain.prompts import PromptTemplate
import warnings
import re
from typing import List, Dict, Any
warnings.filterwarnings("ignore")

class EnhancedCSVAgent:
    def __init__(self, groq_api_key: str, model_name: str = "llama3-8b-8192", summarizer_model: str = "llama3-8b-8192"):
        self.groq_api_key = groq_api_key
        self.model_name = model_name
        self.summarizer_model = summarizer_model
        self.llm = None
        self.summarizer_llm = None
        self.query_enhancer_llm = None
        self.agent = None
        self.df = None
        self.column_info = {}
        self.sample_data = {}
        self.memory = ConversationBufferMemory(memory_key="chat_history", return_messages=True)
        self._setup_llms()

    def _setup_llms(self):
        """Setup all required LLMs"""
        try:
            # Main agent LLM
            self.llm = ChatGroq(
                groq_api_key=self.groq_api_key,
                model_name=self.model_name,
                temperature=0,
                max_tokens=4096,
                streaming=False,
                request_timeout=60
            )
            
            # Summarizer LLM
            self.summarizer_llm = ChatGroq(
                groq_api_key=self.groq_api_key,
                model_name=self.summarizer_model,
                temperature=0.3,
                max_tokens=2048,
                streaming=False,
                request_timeout=30
            )
            
            # Query enhancer LLM
            self.query_enhancer_llm = ChatGroq(
                groq_api_key=self.groq_api_key,
                model_name=self.model_name,
                temperature=0.1,
                max_tokens=1024,
                streaming=False,
                request_timeout=30
            )
            
        except Exception as e:
            raise Exception(f"Error initializing LLMs: {str(e)}")

    def load_csv(self, csv_path: str, verbose: bool = False):
        """Load CSV with enhanced data profiling"""
        try:
            if isinstance(csv_path, str):
                csv_path = [csv_path]

            self.df = pd.read_csv(csv_path[0])
            
            # Convert GPA to numeric if it exists
            if 'GPA' in self.df.columns:
                self.df['GPA'] = pd.to_numeric(self.df['GPA'], errors='coerce')
            
            # Profile the data
            self._profile_data()
            
            # Create enhanced agent with better prompt
            self._create_enhanced_agent(csv_path)
            
            if verbose:
                print(f"📊 Enhanced CSV Agent loaded successfully")
                print(f"   Shape: {self.df.shape}")
                print(f"   Columns: {list(self.df.columns)}")
                print(f"   Sample values for key columns:")
                for col, samples in self.sample_data.items():
                    print(f"     {col}: {samples[:3]}")

        except Exception as e:
            raise Exception(f"Error loading CSV: {str(e)}")

    def _profile_data(self):
        """Create comprehensive data profile"""
        self.column_info = {}
        self.sample_data = {}
        
        for col in self.df.columns:
            # Basic info
            dtype = str(self.df[col].dtype)
            null_count = self.df[col].isnull().sum()
            unique_count = self.df[col].nunique()
            
            # Sample values (non-null, unique)
            sample_values = self.df[col].dropna().unique()[:10].tolist()
            self.sample_data[col] = sample_values
            
            # Determine logical type
            if dtype in ['int64', 'int32', 'int16', 'int8']:
                logical_type = 'integer'
            elif dtype in ['float64', 'float32', 'float16']:
                logical_type = 'float'
            elif dtype == 'object':
                # Check if it's actually numeric
                try:
                    pd.to_numeric(self.df[col].dropna().iloc[:100], errors='raise')
                    logical_type = 'numeric_string'
                except:
                    logical_type = 'text'
            else:
                logical_type = 'other'
            
            self.column_info[col] = {
                'type': logical_type,
                'unique_count': unique_count,
                'null_count': null_count,
                'sample_values': sample_values
            }

    def _create_enhanced_agent(self, csv_path: List[str]):
        """Create agent with enhanced system prompt"""
        
        # Create detailed data context
        data_context = self._generate_data_context()
        
        # Enhanced system prompt - more specific for CSV agent
        system_prompt = f"""
You are an expert data analyst working with CSV data. You have access to a pandas DataFrame through the python_repl_ast tool.

{data_context}

CRITICAL INSTRUCTIONS:
1. ALWAYS use python_repl_ast tool for any data operations
2. The DataFrame is automatically loaded as 'df' 
3. EXACT column names are: {list(self.df.columns)}
4. Always use df.head() first to understand the data structure
5. For string filtering, use: df[df['Column_Name'].str.contains('text', case=False, na=False)]
6. For exact matches, use: df[df['Column_Name'] == 'exact_value']
7. For unique values, use: df['Column_Name'].unique() or df['Column_Name'].drop_duplicates()
8. Always handle NaN values with na=False in string operations

EXAMPLE PATTERNS:
- Find student courses: df[df['Student Name'].str.contains('John', case=False, na=False)]['Course Number'].unique()
- Filter by grade: df[df['Grade'] == 'A']
- Multiple conditions: df[(df['Student Name'].str.contains('John', case=False, na=False)) & (df['Grade'] == 'A')]

Always start by examining the data with df.head() or df.info() if you're unsure about the structure.
        """

        self.agent = create_csv_agent(
            llm=self.llm,
            path=csv_path,
            verbose=True,
            agent_type=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
            allow_dangerous_code=True,
            handle_parsing_errors=True,
            max_iterations=10,  # Reasonable iterations
            max_execution_time=120,  # Reasonable timeout
            return_intermediate_steps=False,
            include_df_in_prompt=False,  # Let agent discover structure
            prefix=system_prompt
        )

    def _generate_data_context(self) -> str:
        """Generate comprehensive data context"""
        context = f"Dataset Shape: {self.df.shape[0]} rows, {self.df.shape[1]} columns\n\n"
        context += "COLUMNS AND SAMPLE DATA:\n"
        
        for col, info in self.column_info.items():
            context += f"- {col} ({info['type']}): "
            context += f"{info['unique_count']} unique values, "
            context += f"Examples: {info['sample_values'][:5]}\n"
        
        return context

    def enhance_query(self, user_query: str) -> str:
        """Use LLM to enhance and clarify user queries"""
        
        enhancement_prompt = f"""
        You are a query enhancement expert for CSV data analysis. Your job is to convert natural language queries into better natural language queries that work with CSV agents.

        AVAILABLE COLUMNS: {list(self.df.columns)}
        
        SAMPLE DATA:
        {self._generate_sample_data_context()}

        USER QUERY: {user_query}

        IMPORTANT: You must return a NATURAL LANGUAGE QUERY, NOT code. The CSV agent will handle the pandas operations.

        Your task:
        1. Identify what the user is asking for
        2. Map their terms to EXACT column names from the available columns
        3. Rephrase the query using the correct column names
        4. Make the query more specific and clear
        5. Return a natural language question that uses the exact column names

        ENHANCEMENT RULES:
        - Always use the EXACT column names from: {list(self.df.columns)}
        - Convert informal names to exact column names
        - Make queries more specific about what data to return
        - For names, be explicit about partial matching needs
        - For filtering, be clear about conditions
        - Return a clear, natural language question

        EXAMPLES:
        Input: "Show me Leslie's courses"
        Output: "What are the unique Course Number values for rows where Student Name contains 'Leslie'?"

        Input: "Students from Murray State"
        Output: "What are the unique Student Name values where Organization equals 'MURRAY STATE COLLEGE'?"

        Return ONLY the enhanced natural language query, nothing else.
        """

        try:
            response = self.query_enhancer_llm.invoke(enhancement_prompt)
            enhanced_query = response.content if hasattr(response, 'content') else str(response)
            # Clean up any code blocks or formatting
            enhanced_query = enhanced_query.strip()
            enhanced_query = enhanced_query.replace('```', '').replace('`', '')
            # Remove any "Here is the enhanced query:" prefixes
            if 'enhanced query:' in enhanced_query.lower():
                enhanced_query = enhanced_query.split('enhanced query:')[-1].strip()
            return enhanced_query
        except Exception as e:
            print(f"⚠️ Query enhancement failed: {e}")
            return user_query

    def _generate_sample_data_context(self) -> str:
        """Generate sample data context for query enhancement"""
        context = ""
        for col, samples in list(self.sample_data.items())[:5]:  # Limit to first 5 columns
            context += f"{col}: {samples[:3]}\n"
        return context

    def _extract_raw_data(self, response: str) -> str:
        """Extract the raw data from agent response"""
        if "Final Answer:" in response:
            return response.split("Final Answer:")[-1].strip()
        return response

    def _summarize_response(self, raw_response: str, original_question: str, format_type: str = "auto") -> str:
        """Enhanced response summarization"""
        
        if format_type == "auto":
            prompt = f"""
            You are an intelligent data presentation expert. Format the following response in the most appropriate way.

            Original Question: {original_question}
            Raw Data Response: {raw_response}

            FORMATTING GUIDELINES:
            1. **TABLE format** for: structured data, multiple records, comparisons, course/student listings
            2. **NARRATIVE format** for: single answers, trends, calculations, summaries
            3. **LIST format** for: simple enumerations, unique values

            RULES:
            - Remove technical jargon and pandas references
            - Use clear, professional language
            - Add context where helpful
            - Use appropriate emojis (📊 📝 📖)
            - Make it user-friendly and engaging
            - If no data found, explain clearly
            - For tables, use proper markdown format

            Choose the best format automatically and present the data clearly.
            Provide only the formatted response.
            """
        else:
            prompt = f"""
            Clean up and format this data response:
            
            Original Question: {original_question}
            Raw Response: {raw_response}
            
            Make it user-friendly by:
            - Removing technical details
            - Using clear formatting
            - Explaining the results
            - Adding context if needed
            
            Provide only the cleaned response.
            """

        try:
            summary_response = self.summarizer_llm.invoke(prompt)
            return summary_response.content if hasattr(summary_response, 'content') else str(summary_response)
        except Exception as e:
            print(f"❌ Summarization failed: {str(e)}")
            return f"Summarization failed. Raw response: {raw_response}"

    def query(self, question: str, enhance_query: bool = True, max_retries: int = 2, 
              format_type: str = "auto", use_summarizer: bool = True):
        """
        Enhanced query method with automatic query improvement
        """
        if not self.agent:
            raise ValueError("No CSV file loaded. Please call load_csv() first.")

        # Always try to understand the data structure first
        print(f"📊 Available columns: {list(self.df.columns)}")
        
        # Enhance the query if requested
        if enhance_query:
            print(f"🔍 Original query: {question}")
            enhanced_question = self.enhance_query(question)
            print(f"✨ Enhanced query: {enhanced_question}")
            question_to_use = enhanced_question
        else:
            question_to_use = question

        # Try with enhanced query first, then fallback to original
        queries_to_try = [question_to_use] if not enhance_query else [question_to_use, question]
        
        for query_attempt, current_question in enumerate(queries_to_try):
            print(f"\n🎯 Query attempt {query_attempt + 1}: {current_question}")
            print("-" * 60)
            
            for retry in range(max_retries):
                try:
                    print(f"🤔 Execution attempt {retry + 1}")
                    
                    # Execute the query
                    response = self.agent.run(current_question)
                    
                    print("✅ Agent completed successfully")
                    
                    if use_summarizer:
                        print("🔄 Formatting response...")
                        raw_data = self._extract_raw_data(response)
                        formatted_response = self._summarize_response(raw_data, question, format_type)
                        print("✅ Formatting completed")
                        return formatted_response
                    else:
                        return self._extract_raw_data(response)
                        
                except Exception as e:
                    print(f"❌ Attempt {retry + 1} failed: {str(e)}")
                    if retry == max_retries - 1:
                        print(f"⚠️ All attempts failed for this query variation")
                    continue
            
            # If we get here, all retries failed for this query variation
            if query_attempt < len(queries_to_try) - 1:
                print(f"🔄 Trying fallback query...")
                continue
        
        return f"❌ Agent failed after trying all query variations and retries. Please rephrase your question or check the data structure."

    def get_data_summary(self) -> str:
        """Get a summary of the loaded data"""
        if self.df is None:
            return "No data loaded"
        
        summary = f"📊 **Data Summary**\n"
        summary += f"- **Rows**: {self.df.shape[0]:,}\n"
        summary += f"- **Columns**: {self.df.shape[1]}\n\n"
        summary += f"**Available Columns:**\n"
        
        for col, info in self.column_info.items():
            summary += f"- **{col}** ({info['type']}): {info['unique_count']} unique values\n"
            if info['sample_values']:
                summary += f"  Examples: {', '.join(map(str, info['sample_values'][:3]))}\n"
        
        return summary

# Example usage with improved error handling
if __name__ == "__main__":
    try:
        # Initialize enhanced agent
        agent = EnhancedCSVAgent(groq_api_key=os.getenv('GROQ_API_KEY'))
        
        # Load CSV
        agent.load_csv("data/csv_folder/student_transcript.csv", verbose=True)
        
        # Show data summary
        print("\n" + "="*60)
        print("DATA SUMMARY")
        print("="*60)
        print(agent.get_data_summary())
        
        # Test queries with various phrasings (more natural language)
        test_queries = [
            "What are the unique Course Number values for student Leslie Nichole Bright?",
            "Show me the Course Number and Term where student Trista Denay Barrett has Grade A",
            "List all unique Student Name values where Organization is MURRAY STATE COLLEGE",
            "Show unique Student Name where Organization is NEWMAN UNIVERSITY",
            "Calculate average GPA for each student and sort in descending order"
        ]
        
        for i, query in enumerate(test_queries[:2], 1):  # Test first 2 queries
            print(f"\n" + "="*60)
            print(f"TEST QUERY {i}")
            print("="*60)
            result = agent.query(query, enhance_query=True, use_summarizer=True)
            print(result)
            
    except Exception as e:
        print(f"❌ Error: {str(e)}")
````
